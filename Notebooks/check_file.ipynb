```python
import os
daily = "../Data/Raw/ERA5/daily"
if os.path.exists(daily):
    files = sorted(f for f in os.listdir(daily) if f.endswith(".nc"))
    print("Files downloaded so far:", len(files))
    for f in files:
        print(f, f"{os.path.getsize(os.path.join(daily, f))/1e3:.0f} KB")
else:
    print("daily/ folder doesn't exist yet — nothing saved")
```
