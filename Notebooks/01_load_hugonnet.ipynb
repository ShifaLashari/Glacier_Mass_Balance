```python
import pandas as pd
import geopandas as gpd
import numpy as np

# --- 1. Load RGI v6 morphology (reads attributes from the .dbf automatically) ---
rgi6 = gpd.read_file(r"C:\DATA\Dissertation\Glacier_Mass\Glacier_Mass_Balance\Glacier_Mass_Balance\Data\Raw\RGI\nsidc0770_11.rgi60.CentralEurope\11_rgi60_CentralEurope.shp")
print("RGI v6 glaciers:", len(rgi6))

# --- 2. Load your saved Hugonnet mass balance ---
hug = pd.read_csv("../Data/Raw/Hugonnet/hugonnet_region11.csv")

# --- 3. Pick ONE Hugonnet period (the full 20-year record) ---
hug20 = hug[hug['period'] == '2000-01-01_2020-01-01'].copy()
print("Hugonnet glaciers (20-yr period):", len(hug20))

# --- 4. THE MERGE — native join, no crosswalk ---
df = hug20.merge(
    rgi6[['RGIId', 'Area', 'Zmin', 'Zmax', 'Zmed', 'Slope', 'Aspect',
           'Lmax', 'CenLon', 'CenLat']],
    left_on='rgiid', right_on='RGIId', how='inner'
)
print("Merged rows:", len(df))

# --- 5. Derive northness / eastness from aspect (degrees -> direction components) ---
df['northness'] = np.cos(np.radians(df['Aspect']))
df['eastness']  = np.sin(np.radians(df['Aspect']))

# --- 6. Quick sanity check ---
print(df[['rgiid', 'dmdtda', 'Area', 'Zmed', 'Slope', 'northness', 'eastness']].head())
print("\nAny missing values?\n", df[['dmdtda','Area','Zmed','Slope','Aspect']].isna().sum())

# --- 7. Save your Stage 1 model-ready table ---
df.to_csv("../Data/Processed/stage1_hugonnet_rgi6_merged.csv", index=False)
print("\nSaved Stage 1 table:", len(df), "glaciers")
```
