```python
import cdsapi
import os

os.makedirs("../Data/Raw/ERA5", exist_ok=True)
c = cdsapi.Client()

out = "../Data/Raw/ERA5/era5_geopotential_alps.nc"

c.retrieve(
    "reanalysis-era5-single-levels",
    {
        "product_type": "reanalysis",
        "variable": "geopotential",
        "year": "2000",
        "month": "01",
        "day": "01",
        "time": "00:00",
        "area": [48, 5, 44, 18],       # same Alps box as before — must match!
        "data_format": "netcdf",
        "download_format": "unarchived",
    },
    out,
)
print("Saved:", out)
```

```python
with open(out, "rb") as f:
    head = f.read(4)
print("First bytes:", head)
print("Size (KB):", round(os.path.getsize(out) / 1e3, 1))
```

```python
import xarray as xr

zg = xr.open_dataset(out)       # or the extracted path if it was zipped
print(zg)                        # confirm variable name is 'z'

# geopotential (m²/s²) -> elevation in metres
z_era5_m = zg["z"].squeeze() / 9.80665

print("ERA5 cell elevations (m):",
      float(z_era5_m.min()), "to", float(z_era5_m.max()))
```
